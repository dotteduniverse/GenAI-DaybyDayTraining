# Day 50: Project Showcase & Presentation

This notebook provides checklists and scripts to help you finalise and submit your capstone project.

```python
# 1. Submission checklist
submission_checklist = """
# Final Submission Checklist

## Code & Documentation
- [ ] GitHub repository is public.
- [ ] README.md includes: project description, setup instructions, usage example, live demo link, blog post link.
- [ ] All code files are committed (no missing days).
- [ ] requirements.txt or environment.yml is present.
- [ ] .env.example file (without real keys) is included.

## Live Demo
- [ ] Hugging Face Space or Replicate endpoint is working.
- [ ] Demo URL is added to README.

## Video
- [ ] Recorded demo (2‑5 min).
- [ ] Video shows: app loading, key feature, edge case handling.
- [ ] Audio is clear (narrate what you're doing).
- [ ] Video uploaded (YouTube unlisted or Loom).

## Presentation (optional but recommended)
- [ ] 5‑10 slides.
- [ ] Cover: title, your name, date.
- [ ] Problem statement.
- [ ] Solution overview + architecture diagram.
- [ ] Live demo embedded (or link).
- [ ] Results / metrics.
- [ ] Lessons learned & future work.
- [ ] Q&A slide.

## Blog Post
- [ ] Published (Medium, Dev.to, etc.).
- [ ] Link in README.

## Final Review
- [ ] Test the app from a fresh environment (clone, install, run).
- [ ] Check for hardcoded secrets.
- [ ] Run through the demo script (no surprises).
"""
print(submission_checklist)
with open("submission_checklist.md", "w") as f:
    f.write(submission_checklist)
print("\nSaved submission_checklist.md")
```

```python
# 2. Demo script template (to follow while recording)
demo_script = """
# Demo Script (2-3 minutes)

0. **Intro** (10 sec): "Hi, I'm [Name]. Today I'll show my project: [Project Name]."

1. **Problem & Solution** (20 sec): "Many people struggle with [problem]. This app solves it by [solution]."

2. **Walkthrough** (60-90 sec):
   - Open the live app.
   - Show the UI (explain inputs).
   - Enter a typical query/upload an image.
   - Click generate/ask.
   - Show the output.
   - Try an edge case (e.g., empty input, long text).
   - Show error handling (if applicable).

3. **Behind the scenes** (30 sec):
   - Mention the model(s), framework, optimisation (vLLM, etc.).
   - Show a code snippet (optional).

4. **Conclusion** (20 sec):
   - Recap what you built.
   - Link to GitHub and blog post.
   - "Thanks for watching!"
"""
print(demo_script)
with open("demo_script.md", "w") as f:
    f.write(demo_script)
print("Saved demo_script.md")
```

```python
# 3. Generate a quick slide outline (can be converted to PowerPoint/Google Slides)
slide_outline = """
# Slide 1: Title
- Project Name
- Your Name
- Date

# Slide 2: Problem
- What pain point does this address?
- Example scenario.

# Slide 3: Solution Overview
- High-level description.
- Screenshot of the app.

# Slide 4: Architecture
- Diagram (e.g., user → UI → API → LLM → response).
- Key components: models, vector store, cache, etc.

# Slide 5: Live Demo
- Embed a short GIF or link to the live app.
- Or record a short clip.

# Slide 6: Results & Metrics
- Response time, cost, accuracy (if measured).
- User feedback summary.

# Slide 7: Lessons Learned
- Biggest challenges (e.g., prompt injection, latency).
- What worked well (e.g., vLLM, FAISS).

# Slide 8: Future Work
- Planned features or improvements.

# Slide 9: Q&A
- Contact info, GitHub, blog.
"""
print(slide_outline)
with open("slide_outline.md", "w") as f:
    f.write(slide_outline)
print("Saved slide_outline.md")
```
