```python
# Exercise 5: Cost tracking
import tiktoken

def count_tokens(text, model="gpt-3.5-turbo"):
    enc = tiktoken.encoding_for_model(model)
    return len(enc.encode(text))

def calculate_cost(prompt, response, model="gpt-3.5-turbo"):
    prompt_tokens = count_tokens(prompt, model)
    response_tokens = count_tokens(response, model)
    # gpt-3.5-turbo pricing: $0.002 per 1K input, $0.002 per 1K output
    cost = (prompt_tokens + response_tokens) * 0.000002
    return cost

print("Example cost:", calculate_cost("What is AI?", "AI stands for Artificial Intelligence."))
```
