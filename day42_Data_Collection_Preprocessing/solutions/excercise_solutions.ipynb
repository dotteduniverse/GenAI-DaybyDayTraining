```python
# Exercise 5: Data augmentation for text (back‑translation with Helsinki-NLP)
# (requires transformers)
from transformers import MarianMTModel, MarianTokenizer

def back_translate(text, src='en', intermediate='fr'):
    # Load models (slow, only for demonstration)
    model_name = f'Helsinki-NLP/opus-mt-{src}-{intermediate}'
    tokenizer = MarianTokenizer.from_pretrained(model_name)
    model = MarianMTModel.from_pretrained(model_name)
    translated = model.generate(**tokenizer(text, return_tensors="pt", padding=True))
    intermediate_text = tokenizer.decode(translated[0], skip_special_tokens=True)
    # Translate back
    back_model = f'Helsinki-NLP/opus-mt-{intermediate}-{src}'
    back_tokenizer = MarianTokenizer.from_pretrained(back_model)
    back_model = MarianMTModel.from_pretrained(back_model)
    back = back_model.generate(**back_tokenizer(intermediate_text, return_tensors="pt", padding=True))
    return back_tokenizer.decode(back[0], skip_special_tokens=True)

# Example: back_translate("The cat sits on the mat.")
```
