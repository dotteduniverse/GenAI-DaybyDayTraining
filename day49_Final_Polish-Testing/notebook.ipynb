```python
# 1. Create a test plan checklist
test_plan = """
# Test Plan for Capstone Project

## Functional Tests
- [ ] Main feature works with valid inputs.
- [ ] Handles empty inputs gracefully.
- [ ] Handles extremely long inputs (truncation or error).
- [ ] All input components (sliders, checkboxes, uploads) work.
- [ ] Output format is as expected.

## Edge Cases
- [ ] Special characters in prompts.
- [ ] Non‑English characters (Unicode).
- [ ] Very large images (if applicable).
- [ ] Network failures (simulate by disabling internet).

## Performance
- [ ] Response time < 10 seconds (or acceptable threshold).
- [ ] No memory leaks after 10 consecutive requests.

## Usability
- [ ] Instructions are clear.
- [ ] Error messages are helpful.
- [ ] UI is responsive on different screen sizes.

## Security (if applicable)
- [ ] API keys are not exposed.
- [ ] Rate limiting works.
- [ ] Prompt injection attempts are blocked.
"""
print(test_plan)

# Save to file
with open("test_plan.md", "w") as f:
    f.write(test_plan)
print("\nSaved test_plan.md")
```
