```python
# Exercise 5: JWT authentication
# pip install pyjwt
import jwt
import datetime

SECRET = "my-secret"

@app.post("/login")
def login(username: str, password: str):
    if username == "user" and password == "pass":
        token = jwt.encode({"user": username, "exp": datetime.datetime.utcnow() + datetime.timedelta(hours=1)}, SECRET)
        return {"token": token}
    raise HTTPException(401, "Invalid credentials")

def decode_token(token: str):
    try:
        return jwt.decode(token, SECRET, algorithms=["HS256"])
    except:
        raise HTTPException(403, "Invalid token")

# Then protect /generate with token in Authorization header.
```
