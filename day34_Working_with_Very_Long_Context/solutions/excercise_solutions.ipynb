```python
# Exercise 5: Manual streaming map‑reduce
def manual_map_reduce(chunks, llm):
    chunk_summaries = []
    for i, chunk in enumerate(chunks):
        summary = llm.invoke(f"Summarise this:\n{chunk}").content
        print(f"Chunk {i+1} summarised.")
        chunk_summaries.append(summary)
    combined = "\n\n".join(chunk_summaries)
    final = llm.invoke(f"Combine into final summary:\n{combined}").content
    return final
```
