# Day 45: API Deployment – Hugging Face Spaces

This notebook guides you through preparing your app for deployment on HF Spaces.

````python
# 1. Create a local folder for your Space
import os
from pathlib import Path

space_dir = Path("./my_hf_space")
space_dir.mkdir(exist_ok=True)
print(f"Created {space_dir}")
````

````python
# 2. Write app.py (Gradio example)
app_py_content = '''
import gradio as gr
import os
from openai import OpenAI

client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

def answer(question):
    response = client.chat.completions.create(
        model="gpt-3.5-turbo",
        messages=[{"role": "user", "content": question}]
    )
    return response.choices[0].message.content

demo = gr.Interface(fn=answer, inputs="text", outputs="text", title="My Capstone")
demo.launch()
'''

with open(space_dir / "app.py", "w") as f:
    f.write(app_py_content)
print("app.py written")
````

````python
# 3. Write requirements.txt
requirements = """
gradio
openai
"""
with open(space_dir / "requirements.txt", "w") as f:
    f.write(requirements.strip())
print("requirements.txt written")
````

````python
# 4. (Optional) Write a README.md for your Space
readme = """
---
title: My Capstone Project
emoji: 🚀
colorFrom: blue
colorTo: purple
sdk: gradio
sdk_version: 4.0.0
app_file: app.py
pinned: false
---

# My Capstone
Ask any question and get an answer from GPT-3.5.
"""
with open(space_dir / "README.md", "w") as f:
    f.write(readme)
print("README.md written")

# Now you can push this folder to a HF Space: https://huggingface.co/new-space
````
