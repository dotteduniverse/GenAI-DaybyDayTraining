```python
# Exercise 3: Token budget
def scrape_and_summarise(url, max_chars=4000):
    text = scrape_website(url)
    if len(text) > max_chars:
        text = llm.invoke(f"Summarise this:\n{text[:max_chars]}").content
    return text
```

```python
# Exercise 5: Multi-query refinement
def iterative_research(query):
    first_pass = research_pipeline(query)
    gaps = llm.invoke(f"What information is missing from this report?\n{first_pass}").content
    second_query = f"{query} specifically: {gaps}"
    second_pass = research_pipeline(second_query)
    final = llm.invoke(f"Combine both reports:\n{first_pass}\n\n{second_pass}").content
    return final
```
