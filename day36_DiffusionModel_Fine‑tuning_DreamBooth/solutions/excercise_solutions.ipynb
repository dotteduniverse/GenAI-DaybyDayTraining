```python
# Exercise 1: Dataset preparation
import cv2
from pathlib import Path

def prepare_images(input_dir, output_dir, size=512):
    output_dir = Path(output_dir)
    output_dir.mkdir(exist_ok=True)
    for img_path in Path(input_dir).glob("*.jpg"):
        img = cv2.imread(str(img_path))
        h, w = img.shape[:2]
        # square crop
        if h > w:
            crop = (h - w) // 2
            img = img[crop:crop+w, :]
        else:
            crop = (w - h) // 2
            img = img[:, crop:crop+h]
        img = cv2.resize(img, (size, size))
        cv2.imwrite(str(output_dir / img_path.name), img)
```
