# Day 41: Project Ideation & Design Doc

This notebook guides you through selecting a capstone project and drafting a design document.

```python
# Helper: brainstorm ideas with LLM
import os
from openai import OpenAI
from dotenv import load_dotenv
load_dotenv()

client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

def brainstorm(interests):
    prompt = f"""Suggest 5 GenAI project ideas for someone interested in: {interests}.
    For each idea, give a title, one sentence description, and the main tech components (models, libraries)."""
    response = client.chat.completions.create(
        model="gpt-4o",
        messages=[{"role": "user", "content": prompt}]
    )
    return response.choices[0].message.content

# Example
print(brainstorm("art, creative writing, automation"))
```

```python
# Template for design doc (can be copied to a file)
design_doc_template = """
# Project Design Document

## Title
[Your project name]

## One‑line description
[What does it do?]

## Problem statement
[What problem does it solve?]

## Target users
[Who will use it?]

## Core features
1. Feature 1
2. Feature 2
3. Feature 3

## Tech stack
- Models: [e.g., GPT-4, CLIP, Stable Diffusion]
- Libraries: [LangChain, FastAPI, Gradio, etc.]
- Deployment: [Hugging Face Spaces, Replicate, etc.]

## Data sources
[If any custom dataset, describe.]

## Evaluation plan
[How will you measure success? e.g., accuracy, user satisfaction, latency]

## Timeline (days 42–50)
- Day 42: Data collection & preprocessing
- Day 43: Prototype UI
- Day 44: Optimisation (vLLM)
- Day 45: API deployment
- Day 46: Monitoring
- Day 47: Security & rate limiting
- Day 48: Blog post
- Day 49: Final polish & testing
- Day 50: Showcase & presentation
"""
print(design_doc_template)

# Save to file
with open("DESIGN.md", "w") as f:
    f.write(design_doc_template)
print("\nDESIGN.md created – edit with your project details.")
```
